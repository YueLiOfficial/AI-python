```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
data = np.random.randint(50, 101, 10)
scores = pd.Series(data, index = [f'学生{i}' for i in range(1, 11)])
print(scores.mean())
print(scores.max())
print(scores.min())
print((scores[scores  > scores.mean()]).count())
```
