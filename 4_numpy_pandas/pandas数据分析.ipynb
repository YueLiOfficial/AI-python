## 缺失值处理

```python
s = pd.Series([1, 2, np.nan, None, pd.NA])
df = pd.DataFrame([[1, pd.NA, 2], [2, 3, 5], [np.nan, 4, 6]], columns=['第一列', '第二列', '第三列'])

print(s.isna())
print(s.isna().sum())

print(df.isna())
print(df.isna().sum())
```

```python
# 删除缺失值
print(s.dropna())

print(df.dropna()) # 删除带缺失值的一整条记录
print(df.dropna(how='all')) # 当这一条记录中都是缺失值才删除
print(df.dropna(thresh=2)) # 如果至少有n个值不是缺失值，就保留
print(df.dropna(axis=1)) # 删除一列的记录
print(df.dropna(subset=['第一列'])) # 指定查找某一列如果有缺失值，就删除一整行
```

```python
# 填充缺失值

df = pd.read_csv('data/weather_withna.csv')

# df.tail()
print(df.isna().sum())

# 使用字典填充
print(df.fillna({'temp_max': 20, 'wind': 2.5}).tail())

# 用平均值填充
print(df.fillna(df[['wind']].mean()).tail())

# 用前面相邻的值填充
print(df.ffill().tail())

# 用后面相邻的值填充
print(df.bfill().tail())
```

## 清洗错误数据

```python
person = {
  "name": ['Google', 'Runoob' , 'Taobao'],
  "age": [50, 40, 12345]    # 12345 年龄数据是错误的
}

df = pd.DataFrame(person)

df.loc[2, 'age'] = 30
print(df.to_string())

# 条件语句清洗数据
person = {
  "name": ['Google', 'Runoob' , 'Taobao'],
  "age": [50, 200, 12345]    
}

df = pd.DataFrame(person)

for x in df.index:
    if df.loc[x, 'age'] > 120:
        df.loc[x, 'age'] = 120
df
```

## 数据分箱

```python
# pd.cut(data, bins, labels) bins:分箱的范围, labels: 分箱后范围的标签
df = pd.read_csv('data/employees.csv')
df1 = df[['employee_id', 'salary']].head(10)

df['收入级别'] = pd.cut(df['salary'], 
       bins=[0, 10000, 20000, 30000], 
       labels=['低', '中', '高'])
print(df['收入级别'].value_counts().to_string())

# pd.qcut() 等频分箱
print(pd.qcut(df1['salary'], q=3, labels=['低', '中', '高']).value_counts().to_string())
```

## 时间处理

```python
import pandas as pd

d = pd.Timestamp('2026-08-17 22:10:03')

print(type(d))

print(d.year, d.month, d.day, d.hour, d.minute, d.second)
print('季度: ', d.quarter)
print('是否为月底: ', d.is_month_end)

# 方法
print('星期几: ', d.day_name())
print('转换为天: ', d.to_period('D'))
print('转换为月: ', d.to_period('M'))
print('转换为年: ', d.to_period('Y'))
print('转化为季度: ', d.to_period('Q'))
print('转换为周: ', d.to_period('W'))
```

## 分组聚合

```python
# df.groupby(分组字段)[聚合字段].聚合函数()

df = pd.read_csv('data/employees.csv')

# 删除缺失值
df = df.dropna(subset=['department_id'])

df.groupby('department_id').groups # 查看所有分组
df.groupby('department_id').get_group(10) # 查看某个分组的数据

df1 = df.groupby('department_id')[['salary']].mean().round(2)
df1 = df1.reset_index()
df1.sort_values('salary', ascending=False)
```

```python
# 两个字段聚合
df2 = df.groupby(['department_id', 'job_id'])[['salary']].mean()
df2 = df2.reset_index()
df2.sort_values('salary', ascending=False)
```
