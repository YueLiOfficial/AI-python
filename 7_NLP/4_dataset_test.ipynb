```python
from transformers import AutoModel, AutoTokenizer
from datasets import load_dataset
```

```python
tokenizer = AutoTokenizer.from_pretrained("./data/bert-base-chinese")
```

```python
dataset_dict = load_dataset("csv", data_files="./data/online_shopping_10_cats.csv")
```

```python
dataset = dataset_dict["train"]
```

```python
print(dataset)
```

```python
dataset = dataset.remove_columns("cat")
print(dataset)
```

```python
dataset = dataset.filter(lambda x: x["review"] is not None and x["review"].strip() != "" and x["label"] in [0, 1])
print(dataset)
```

```python
tokenizer(list(dataset[:10]["review"]), padding=True, truncation=True, max_length=10)
```

```python
def tokenize(example):
    encoded =  tokenizer(
        example["review"],
        padding="max_length",
        truncation=True,
        max_length=128
    )
    example['input_ids'] = encoded['input_ids']
    example['attention_mask'] = encoded['attention_mask']
    
    return example
```

```python
dataset = dataset.map(tokenize, batched=True)
```

```python
print(dataset)
```
