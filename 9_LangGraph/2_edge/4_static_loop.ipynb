```python
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import MessagesState
from langchain.chat_models import init_chat_model
from langchain.tools import tool
from langchain.messages import HumanMessage, AIMessage, SystemMessage, ToolMessage

from random import randint
from typing import Literal
from dotenv import load_dotenv
from IPython.display import display
from rich import print as rprint

load_dotenv(override=True)

llm = init_chat_model(
    model="deepseek-flash",
    model_provider="deepseek",
    extra_body={
        "thinking": {
            "type": "disabled"
        }
    }
)

@tool(parse_docstring=True)
def get_weather(city: str = "北京"):
    """
    Get the current weather for a given city.

    Args:
        city: The name of the city to get the weather for.
    """

    return f"{city} 今天天气晴朗，东南风三级，气温 25~30 ℃"

@tool(parse_docstring=True)
def get_news(domain: Literal["AI", "食品安全"]):
    """
    Get the latest news for a given domain.

    Args:
        domain: The domain to get the news for.
    """

    if domain == "AI":
        return "Anthropic 发布了 Claude Opus-4.8，但通过 API 用中文向它发送“你是谁？”时，大多数情况下返回的却是“Qwen”或“Deepseek”。"
    else:
        return "双汇发展子公司猪肉产品被抽检出抗生素超标37.5倍"

tools = [get_weather, get_news]
llm = llm.bind_tools(tools)

class OverAllState(MessagesState):
    user_input: str
    final_answer: str

def input_node(state: OverAllState) -> OverAllState:
    user_input = state["user_input"]

    return {
        "messages": [(HumanMessage(content=user_input))]
    }

def llm_node(state: OverAllState) -> OverAllState:
    messages = state["messages"]

    res = llm.invoke(messages)

    return {
        "messages": [res]
    }

def tool_node(state: OverAllState) -> OverAllState:
    messages = state["messages"]
    ai_msg = messages[-1]

    tool_calls = ai_msg.tool_calls

    # faild_prob代表失败的概率是0.6
    faild_prob = 6

    for tool_call in tool_calls:
        if tool_call["name"] == "get_weather":
            prob = randint(0, 9)
            if prob < faild_prob:
                messages.append(ToolMessage(content="网络波动，调用工具失败，请重新调用", tool_call_id=tool_call["id"]))
            else:
                tool_res = get_weather.invoke(tool_call)
                messages.append(tool_res)

        elif tool_call["name"] == "get_news":
            prob = randint(0, 9)
            if prob < faild_prob:
                messages.append(ToolMessage(content="网络波动，工具调用失败，请重新调用", tool_call_id=tool_call["id"]))
            else:
                tool_res = get_news.invoke(tool_call)
                messages.append(tool_res)

    return {
        "messages": messages
    }

def output_node(state: OverAllState) -> OverAllState:
    return {
        "final_answer": state["messages"][-1].content
    }

def router(state: OverAllState) ->Literal["tool_node", "output_node"]:
    messages = state["messages"]

    if messages[-1].tool_calls:
        return "tool_node"

    return "output_node"

builder = StateGraph(state_schema=OverAllState)

builder.add_node(input_node)
builder.add_node(llm_node)
builder.add_node(tool_node)
builder.add_node(output_node)

builder.add_edge(START, "input_node")
builder.add_edge("input_node", "llm_node")
builder.add_conditional_edges("llm_node", router)
builder.add_edge("tool_node", "llm_node")
builder.add_edge("output_node", END)

graph = builder.compile()

res = graph.invoke({
    "user_input": "帮我查询杭州市的天气和AI热点",
    "messages": [SystemMessage(content="你是一个AI助手，请调用工具帮我查询，如果工具调用失败则重新调用工具直到调用成功")]
})

print('=' * 30, '-> ai_res <-', '=' * 30)
print("user_input: ", res["user_input"])
print("final_answer: ", res["final_answer"])
for msg in res["messages"]:
    rprint(msg)

display(graph)
```
