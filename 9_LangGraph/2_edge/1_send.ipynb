```python
from langgraph.graph import START, END, StateGraph
from langgraph.types import Send
from langchain.chat_models import init_chat_model
from typing import TypedDict, Annotated, Literal, Sequence
from operator import add
from IPython.display import display
```

```python
class State(TypedDict):
    tasks: list[str]
    outputs: Annotated[list[str], add]

class PrivateState(TypedDict):
    task: str

class InputState(TypedDict):
    tasks: list[str]

def worker_node(state: PrivateState) -> State:
    return {
        "outputs": [f"{state['task']}执行完毕"]
    }

def send_router(state: InputState) -> Sequence[Send]:
    return [
        Send(
            "worker_node",
            {
                "task": task
            }
        ) for task in state["tasks"]
    ]

builder = StateGraph(
    state_schema=State,
)

builder.add_node(worker_node)
builder.add_conditional_edges(
    START,
    send_router,
    path_map=["worker_node"]
)
builder.add_edge("worker_node", END)

graph = builder.compile()

tasks = ["task_1", "task_2", "task_3"]

res = graph.invoke({"tasks": tasks})

print(res)

display(graph)
```
