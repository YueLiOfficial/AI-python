```python
from langgraph.graph import START, END, StateGraph
from langgraph.types import Command
from typing import TypedDict, Annotated, Literal
from operator import add
```

```python
class State(TypedDict):
    inputs: Literal["weather", "calculate", "order"]
    outputs: Annotated[list[str], add]

def router_node(state: State) -> Command:
    type = state["inputs"]

    if type == "weather":
        return Command(
            goto="weather_node"
        )
    elif type == "calculate":
        return Command(
            goto="calculate_node"
        )
    elif type == "order":
        return Command(
            goto="order_node"
        )
    else:
        return Command(
            goto=END
        )

def weather_node(state: State) -> State:
    return {
        "outputs": ["这是weather节点"]
    }

def calculate_node(state: State) -> State:
    return {
        "outputs": ["这是calculate节点"]
    }

def order_node(state: State) -> State:
    return {
        "outputs": ["这是order节点"]
    }

builder = StateGraph(state_schema=State)

builder.add_node(weather_node)
builder.add_node(calculate_node)
builder.add_node(order_node)
builder.add_node(router_node)

builder.add_edge(START, "router_node")
builder.add_edge("weather_node", END)
builder.add_edge("calculate_node", END)
builder.add_edge("order_node", END)

graph = builder.compile()

res = graph.invoke({"inputs": "other"})

print(res)
```
