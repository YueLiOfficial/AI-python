```python
import pandas as pd
```

```python
data = {
    '姓名': ['张三', '李四', '王五', '赵六', '钱七'],
    '数学': [85, 92, 78, 88, 95],
    '英语': [90, 88, 85, 92, 80],
    '物理': [75, 80, 88, 85, 90]
}

df = pd.DataFrame(data)

df['总分'] = df[['数学', '英语', '物理']].sum(axis=1)
df['平均分'] = df['总分'] / 3

print(df[(df['数学'] > 90) | (df['英语'] > 85)])

print(df.sort_values('总分', ascending=False).head(3))
```

```python
data = {
    '产品名称': ['A', 'B', 'C', 'D'],
    '单价': [100, 150, 200, 120],
    '销量': [50, 30, 20, 40]
}

df = pd.DataFrame(data)

df['销售额'] = df['单价'] * df['销量']

df.nlargest(1, columns='销售额')

df.sort_values('销售额', ascending=False)
```

```python
data = {
    '用户ID': [101, 102, 103, 104, 105],
    '用户名': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    '商品类别': ['电子产品', '服饰', '电子产品', '家居', '服饰'],
    '商品单价': [1200, 300, 800, 150, 200],
    '购买数量': [1, 3, 2, 5, 4]
}

df = pd.DataFrame(data)

df['消费金额'] = df['商品单价'] * df['购买数量']

df.nlargest(1, columns='消费金额')

# print(df['消费金额'].mean().round(2))
print(f"{df['消费金额'].mean():.2f}")

df[df['商品类别'] == '电子产品']['购买数量'].sum()
```
