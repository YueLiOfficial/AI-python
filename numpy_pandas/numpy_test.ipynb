```python
import numpy as np

temps = np.array([28, 30, 29, 31, 32, 30, 29])
print(temps > 30)
```

```python
arr = np.random.randint(1, 10, (3, 4))
print(arr)

print(np.max(arr, axis=0))
print(np.min(arr, axis=1))

print(np.where(arr % 2, -1, arr))
```

```python
arr = np.arange(1, 13, 1)
print(arr)
arr = np.reshape(arr, (3, 4))
print(arr)
print(np.sum(arr, axis=1))
print(np.mean(arr, axis=0))
print(np.reshape(arr, (12)))
```

```python
np.random.seed(0)
arr = np.random.randint(0, 20, (5, 5))
print(arr)
print(arr[arr > 10])
# arr[arr > 10] = 0
# print(arr)
print(np.where(arr > 10, 0, arr))
```

```python
arr = np.array([120, 135, 110, 125, 130, 140])
print(np.sum(arr))
print(np.mean(arr).round(2))
print(np.var(arr).round(2))

print(np.argmax(arr) + 1)
print(np.argmin(arr)  +1)
```

```python
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

print(np.concatenate((A, B)))
print(np.vstack((A, B))) # 纵向拼接
print(np.hstack((A, B))) # 横向拼接
```

```python
arr = np.array([2, 1, 2, 3, 1, 4, 3])
u_arr, counts = np.unique(arr, return_counts=True)
print(u_arr)
print(counts)
```

```python
arr1 = np.array([20, 25, 22, 30, 28])
arr2 = np.array([15, 18, 16, 22, 20])

arr3 = arr1 - arr2
print(arr3)

print(np.mean(arr3).round(2))
print(np.std(arr3).round(2))

# print(np.size(arr3[arr3 ==np.max(arr3)]))
print(np.where(arr3 == np.max(arr3)))
```
