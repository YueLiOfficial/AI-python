## series的创建

```python
import pandas as pd

s = pd.Series([1, 2, 3, 4, 5], index = ['a', 'b', 'c', 'd', 'e'], name='A')
print(s)

s1 = pd.Series(s, index = ['a', 'c'])
print(s1)
```

## series的属性

```python
# 获取索引对象
print(s.index)
# 获取值
print(s.values)
# 获取元素类型、形状、维度、元素个数、名称
print(s.dtype, s.shape, s.ndim, s.size, s.name)
```

## 元素访问

```python
print(s.loc['a']) # 标签索引（显式索引）
print(s.loc['a': 'c'])
print(s.iloc[0]) # 位置索引（隐式索引）
print(s.at['a']) # at无法切片
print(s.iat[0])
# 布尔切片
print(s[s > 2])
```

## 常见函数

```python
s = pd.Series([1, 2, None, None, 3, 4, 5], index = ['A', 'B', 'C', 'D', 'E', 'F', 'G'], name='data')
print(s)
```

```python
s.head(3) # 默认取前五行数据
s.tail() # 默认取后五行数据
```

```python
s.describe() # 查看所有描述性信息
```

```python
print(s.count()) # 获取元素个数（忽略缺失值）
```

```python
# 获取索引
print(s.keys()) # 通过方法获取
print(s.index) #  通过属性获取
```

```python
# 判断缺失值
print(s.isna())
# 判断非缺失
print(s.notna())
```

```python
s.isin([4, 5]) # 检查每个元素是否在某个集合中
```

```python
print(s.var())
print(s.std())
print(s.min())
print(s.max())
print(s.sum())
print(s.median())
print(s.mean())
print(s.diff()) # 求相邻元素差值
```

```python
print(s.sort_values())
print(s.quantile(0.8))
```

```python
s['H'] = 3
s.mode() # 获取众数
```

```python
print(s.value_counts()) # 每个元素出现的次数
print(s.unique()) # 去重
print(s.drop_duplicates()) # 去重返回series类型
print(s.nunique()) # 去重后还有几个元素
print(s.dropna()) # 去除缺失值
```

```python
# 排序
print(s.sort_values()) # 按值排序
print(s.sort_index()) # 按索引排序
```
