## 创建DataFrame

```python
import pandas as pd
```

```python
# 直接创建
data = [['Google', 10], ['Runnoob', 12], ['Wiki', 13]]
df = pd.DataFrame(data, columns=['Site', 'Age'])
#设置数据类型
df['Site'] = df['Site'].astype(str)
df['Age'] = df['Age'].astype(float)
print(df)

# 使用字典创建
data = {'Site': ['Google', 'Runoob', 'Wiki'], 'Age': [10, 12, 13]}
df = pd.DataFrame(data)
print(df)

data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

df = pd.DataFrame(data, index=['day1', 'day2', 'day3'])
print(df)
print(df.loc[['day1', 'day2']])
```

```python
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
df = pd.DataFrame(data, index=['day1', 'day2', 'day3'])

print(df.info())
print(df.describe())
print(df.shape)
print(df.columns)
print(df.index)
print(df.dtypes)
```

```python
print(df.sort_values('calories', ascending=False)) # 按照指定列排序
print(df.sort_index(ascending=False)) # 按行索引排序

data = {
  "calories": [420, 380, None],
  "duration": [50, 40, 45]
}
df = pd.DataFrame(data, index=['day1', 'day2', 'day3'])
print(df.dropna(axis=1)) # 删除含有缺失值的行或列
```

```python
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
df = pd.DataFrame(data, index=['day1', 'day2', 'day3'])
print(df.calories) # 访问某一列
print(df["calories"]) # 访问某一列
df.loc['day4'] = [350, 40] # 按行索引访问某行
print(df)
new_row = pd.DataFrame([[300, 35]], columns=['calories', 'duration'], index = ['day5'])
df = pd.concat([df, new_row])
print(df)
```

```python
print(df)
df_reset = df.reset_index(drop=True) #重置索引
print(df_reset)
df_set = df.set_index('calories') # 将指定列设置为索引
print(df_set)
```

```python
print(df.drop('calories', axis=1)) #删除列
# 也可以直接用del删除列
del df['calories']
print(df.drop('day1')) # 删除行
```

```python
print(df[df['calories'] > 350])
```

```python
# 纵向拼接
df1 = pd.DataFrame({
    'name': ['张三', '李四'],
    'score': [90, 85]
})

df2 = pd.DataFrame({
    'name': ['王五', '赵六'],
    'score': [88, 92]
})

print(pd.concat([df1, df2], ignore_index=True))

# 横向拼接，根据某个字段进行关联
df1 = pd.DataFrame({
    'id': [1, 2, 3],
    'name': ['张三', '李四', '王五']
})
df2 = pd.DataFrame({
    'id': [1, 2, 3],
    'score': [60, 80, 50]
})

print(pd.merge(df1, df2, on='id'))

# 长格式变为宽格式，index指定行，columns指定列，values指定分数
df = pd.DataFrame({
    'name': ['张三', '张三', '李四', '李四'],
    'subject': ['数学', '英语', '数学', '英语'],
    'score': [90, 80, 85, 95]
})
print(df.pivot(index='name',
               columns='subject',
               values='score'
               ))

# 宽格式变为长格式
df = pd.DataFrame({
    'name': ['张三', '李四'],
    '数学': [90, 85],
    '英语': [80, 95]
})
print(df.melt(
    id_vars='name',
    value_vars=['数学', '英语']
))
```
