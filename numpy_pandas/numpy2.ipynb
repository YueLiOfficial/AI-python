```python
import numpy as np
```

### 创建普通数组

```python
arr = np.array([1, 2,3 ])
print(arr)
```

### 创建全空数组

```python
arr = np.empty((2, 3), dtype=int)
print(arr)
```

### 创建全0数组

```python
arr = np.zeros((2, 3))
print(arr)
```

### 创建全1数组

```python
arr = np.ones((2, 3), dtype=int)
print(arr)
print(arr.dtype)
print(arr.shape)
```

### 按照值填充


```python
print(np.full((2, 3), 2))
```

### 创建给定步长的数组

```python
arr = np.arange(1, 11, 1)
print(arr)
```

### 创建等间隔的数组

```python
arr = np.linspace(1, 10, 4)
print(arr)
```

### 创建等比数列

```python
arr = np.logspace(1, 10, 4, base=2)
print(arr)
```

### 特殊矩阵

```python
# 单位矩阵:主对角线上全为1，其余元素全为0
arr = np.eye(3)
print(arr)

# 对角矩阵：只有主对角线的元素不为0，其余元素全为0
print(np.diag([1, 3, 5]))
```

### 随机生成数组

```python
# 随机生成(0,1)之间的浮点数
arr = np.random.rand(2, 3)
print(arr)

# 随机生成给定范围内的浮点数
arr = np.random.uniform(1, 10, (2, 3))
print(arr)

# 随机生成给定范围内的整数
arr = np.random.randint(1, 30, (2, 3))
print(arr)

# 随机生成正态分布的数
arr = np.random.randn(2, 3)
print(arr)

# 设置随机种子
np.random.seed(20)
arr = np.random.randint(1, 20, (2, 3))
print(arr)
```
